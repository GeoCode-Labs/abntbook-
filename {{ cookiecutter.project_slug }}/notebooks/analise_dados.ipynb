# Análise de Dados

Este notebook demonstra técnicas básicas de análise de dados.

```python
# Importar bibliotecas
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar estilo
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

## Geração de Dados de Exemplo

```python
# Gerar dados sintéticos
np.random.seed(42)
n = 100

dados = pd.DataFrame({
    'x': np.random.randn(n),
    'y': np.random.randn(n),
    'categoria': np.random.choice(['A', 'B', 'C'], n)
})

# Adicionar coluna derivada
dados['z'] = dados['x'] * 2 + dados['y'] + np.random.randn(n) * 0.5

dados.head()
```

## Estatísticas Descritivas

```python
# Estatísticas descritivas
dados.describe()
```

## Visualizações

```python
# Gráfico de dispersão
plt.figure(figsize=(10, 6))
sns.scatterplot(data=dados, x='x', y='z', hue='categoria', s=100, alpha=0.6)
plt.title('Relação entre X e Z por Categoria')
plt.xlabel('X')
plt.ylabel('Z')
plt.legend(title='Categoria')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Histograma
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].hist(dados['x'], bins=20, edgecolor='black', alpha=0.7)
axes[0].set_title('Distribuição de X')
axes[0].set_xlabel('X')
axes[0].set_ylabel('Frequência')

axes[1].hist(dados['y'], bins=20, edgecolor='black', alpha=0.7, color='orange')
axes[1].set_title('Distribuição de Y')
axes[1].set_xlabel('Y')
axes[1].set_ylabel('Frequência')

axes[2].hist(dados['z'], bins=20, edgecolor='black', alpha=0.7, color='green')
axes[2].set_title('Distribuição de Z')
axes[2].set_xlabel('Z')
axes[2].set_ylabel('Frequência')

plt.tight_layout()
plt.show()
```

```python
# Boxplot por categoria
plt.figure(figsize=(10, 6))
sns.boxplot(data=dados, x='categoria', y='z')
plt.title('Distribuição de Z por Categoria')
plt.xlabel('Categoria')
plt.ylabel('Z')
plt.show()
```

## Análise de Correlação

```python
# Matriz de correlação
correlacao = dados[['x', 'y', 'z']].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlacao, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Matriz de Correlação')
plt.show()
```
